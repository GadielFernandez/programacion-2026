```python
#Situación: calculadora de propina inteligente.

try:
    cuenta = float(input("ingrese total de la cuenta: "))
    porcentaje = float(input("ingrese porcentaje de propina: "))

    propina = cuenta * porcentaje / 100
    total = cuenta + propina

    mensaje = f"la propina es: ${propina:.2f}. el total a pagar es: ${total:.2f}"

except ValueError:
    mensaje = "Error: debe de ingresar números válidos."

print(mensaje)
```

```python
# Situación: Validador de compra con descuento

try:
    cliente = input("ingrese nombre del cliente: ").upper().strip()
    monto = float(input("ingrese el monto: ").strip())
    frecuente = input("¿es cliente frecuente? (S/N): ").strip()=="S"

    if monto < 0:
        mensaje = "Error: monto inválido"
    else:
        if monto >= 300:
            porcentaje = 15
        elif monto >= 100 and frecuente:
            porcentaje = 20
        elif monto >= 100 or frecuente:
            porcentaje = 10
        else:
            porcentaje = 0

        monto_descuento = monto * porcentaje / 100
        monto_final = monto - monto_descuento

        mensaje = f"cliente: {cliente}. El monto del descuento: ${monto_descuento:.2f}. Monto final: ${monto_final:.2f}"


except ValueError:
    mensaje = "Error: el monto debe de ser un número"

print(mensaje)
```

```python
#Situación: detector de número par e impar.

try:
    número = int(input("ingrese un número (0-100): "))

    if número < 0 or número > 100:
        mensaje = "Error: el número debe de estar entre 0-100"
    elif número % 2 == 0:
        mensaje = f"el número: {número} es par."
    else:
        mensaje = f"el número {número} es impar"

except ValueError:
    mensaje = "Error: debe ingresar un número entero."

print(mensaje)
```

detector_de_numero_par_e_impar

```python
# sitiuación: Clasificador de edad

try:
    edad = int(input("ingrese su edad: ").strip())

    if edad <= 0:
        mensaje = "Error: edad invalida"
    elif edad >= 65:
        mensaje = "adulto mayor"
    elif edad >= 40:
        mensaje = "señor"
    elif edad >= 18:
        mensaje = "mayor de edad"
    elif edad >= 13:
        mensaje = "adolescente"
    else:
        mensaje = "niño"

except ValueError:
    mensaje = "Error: datos inválidos"

print(mensaje)
```
